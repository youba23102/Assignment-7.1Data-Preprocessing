```python
import pandas as pd      #here importing pandas specialize in datasets
import numpy as np       #import numpy library specialse in math 
```

```python
path = r"C:\Users\ayoub\OneDrive\Desktop\machine learning ai\assigment 7.1\Canada by Citizenship.csv"
df = pd.read_csv(path,skiprows=20,na_values="..")  # import the the path of the dataset and make it in a vaiable called "df"
df.head() #this function make us see the first lines of the dataset and its features
# skiprows=20,na_values=".." means skipp the first 20 rows and dont make in the data framke because its has string and
#it will be null in the data fram
```

```python
# Drop columns we don't need
df.drop(['Type', 'Coverage', 'AREA', 'REG', 'DEV'], axis=1, inplace=True)
#drop means deleate and we give it colums name we wamt to delete and axis 1 means columns only
#inplace true means make modification on the same dataset df it self , we dont have to assigne another name

# Rename columns
df.rename(columns={
    'OdName'   : 'Country',
    'AreaName' : 'Continent',
    'RegName'  : 'Region'
}, inplace=True)

# here we rename the colums we still have as the same the example we have 

# Set Country as the index
df.set_index('Country', inplace=True)
#in this step we make the column country the label for our data set

#Add Total column 
df['Total'] = df.loc[:, '1980':'2013'].sum(axis=1)
#in this last step we are making a new column "total" and we are adding all number for each country fro 1980 to 2013 
#and make the total fo each country

df.head()  #here we are printing the result
```
